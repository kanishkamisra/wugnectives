```python
import utils
import re
```

```python
stimuli = utils.read_csv_dict("../data/stimuli-nonce/all_prompts.csv")

temporal = []
for item in stimuli:
    if item['stimuli_type'] == "temporal":
        temporal.append(item)
```

```python
temporal_position_annotations = []
for item in temporal:
    e1, e2 = item['prompt'].find(item['entity1']), item['prompt'].find(item['entity2'])
    if e1 < e2:
        item['first'] = item['entity1']
        item['second'] = item['entity2']
    else:
        item['first'] = item['entity2']
        item['second'] = item['entity1']

    temporal_position_annotations.append(item)
```

```python
utils.write_dict_list_to_csv(temporal_position_annotations, "../data/stimuli-nonce/temporal-position-annotations.csv")
```
