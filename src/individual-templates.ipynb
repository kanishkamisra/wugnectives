```python
import re
import utils
from collections import defaultdict
```

```python
preference_stimuli = utils.read_csv_dict("../data/stimuli-nonce/preference_stimuli.csv")
instantiation_stimuli = utils.read_csv_dict("../data/stimuli-nonce/instantiation_stimuli.csv")
temporal_stimuli = utils.read_csv_dict("../data/stimuli-nonce/temporal_stimuli.csv")

stimuli = {
    "preference": preference_stimuli,
    "instantiation": instantiation_stimuli,
    "temporal": temporal_stimuli
}
```

```python
def templatify(entry):
    premise = entry['premise'].replace(entry['entity1'], "[e1]").replace(entry['entity2'], "[e2]")
    premise = re.sub(r'(hate)\s(leafy vegetables|mammals|fruits|string instruments|insects|snowy winters|living near mountains|warm weather|coastal cities|tall buildings)', "hate P", premise)
    premise = re.sub(r'(love)\s(leafy vegetables|mammals|fruits|string instruments|insects|snowy winters|living near mountains|warm weather|coastal cities|tall buildings)', "love P", premise)
    return premise
```

```python
connectives = defaultdict(set)

for k, v in stimuli.items():
    for entry in v:
        template = templatify(entry)
        connectives[entry['connective']].add((k, template))
```

```python
len(connectives)
```

```python
unique_template = []
for k, v in connectives.items():
    for stimuli_type, vv in v:
        unique_template.append((k,stimuli_type, vv))
```

```python
unique_template

utils.write_csv(unique_template, "../data/unique_templates.csv", header=["connective", "stimuli_type", "template"])
```
